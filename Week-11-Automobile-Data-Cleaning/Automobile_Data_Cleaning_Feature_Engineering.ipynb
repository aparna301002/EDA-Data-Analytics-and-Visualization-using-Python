# **Week 11 – Automobile Data Understanding, Cleaning and Feature Engineering**

Step 1: Import required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings

from google.colab import files

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

Step 2: Upload and load the automobile dataset

```python
df = pd.read_csv("/content/drive/MyDrive/EDA-Skill/Automobile_data.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

Step 3: Understand the dataset

```python
print("First five records:")
display(df.head())

print("\nLast five records:")
display(df.tail())

print("\nDataset dimensions:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").T)
```

Step 4: Standardize column names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace("-", "_", regex=False)
    .str.replace(" ", "_", regex=False)
)

print("Standardized column names:")
print(df.columns.tolist())
```

Step 5: Replace missing value symbols

```python
# Replace question marks and blank strings with NaN
df = df.replace(r"^\s*\?\s*$", np.nan, regex=True)

df = df.replace(r"^\s*$", np.nan, regex=True)

print("Missing values after replacement:")
display(df.isnull().sum().to_frame("Missing Count"))

print("\nMissing value percentages:")
missing_percentage = (
    df.isnull().mean() * 100
).round(2)

display(
    missing_percentage.to_frame("Missing Percentage")
)
```

Step 6: Identify duplicate records

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(df.shape)
```

Step 7: Convert numerical columns

```python
numerical_columns = [
    "wheel_base",
    "length",
    "width",
    "height",
    "curb_weight",
    "engine_size",
    "bore",
    "stroke",
    "compression_ratio",
    "horsepower",
    "peak_rpm",
    "city_mpg",
    "highway_mpg",
    "price"
]

# Convert only columns available in the dataset
numerical_columns = [
    col for col in numerical_columns
    if col in df.columns
]

for col in numerical_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

print("Numerical columns converted successfully.")

display(df[numerical_columns].dtypes)
```

Step 8: Analyze missing values

```python
missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

display(
    missing_summary.sort_values(
        by="Missing_Count",
        ascending=False
    )
)
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False
)

plt.title("Missing Values in Automobile Dataset")
plt.xlabel("Dataset Columns")
plt.show()
```

Step 9: Handle invalid zero values

```python
zero_invalid_columns = [
    "wheel_base",
    "length",
    "width",
    "height",
    "curb_weight",
    "engine_size",
    "bore",
    "stroke",
    "horsepower",
    "peak_rpm",
    "city_mpg",
    "highway_mpg",
    "price"
]

zero_invalid_columns = [
    col for col in zero_invalid_columns
    if col in df.columns
]

for col in zero_invalid_columns:
    df.loc[df[col] <= 0, col] = np.nan

print("Invalid non-positive values handled.")

display(df[zero_invalid_columns].isnull().sum())
```

Step 10: Handle missing numerical values using manufacturer-wise median

```python
if "make" in df.columns:
    df["make"] = df["make"].astype("string").str.strip()

for col in numerical_columns:

    if df[col].isnull().sum() > 0:

        if "make" in df.columns:
            manufacturer_median = (
                df.groupby("make")[col]
                .transform("median")
            )

            df[col] = df[col].fillna(
                manufacturer_median
            )

        # Fallback to overall median
        df[col] = df[col].fillna(
            df[col].median()
        )

print("Numerical missing values handled.")

display(df[numerical_columns].isnull().sum())
```

Step 11: Handle missing categorical values

```python
categorical_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

for col in categorical_columns:

    df[col] = df[col].astype("string").str.strip()

    if df[col].isnull().sum() > 0:

        mode_values = df[col].mode()

        if not mode_values.empty:
            df[col] = df[col].fillna(
                mode_values.iloc[0]
            )
        else:
            df[col] = df[col].fillna("Unknown")

print("Categorical missing values handled.")

display(df[categorical_columns].isnull().sum())
```

Step 12: Validate the cleaned dataset

```python
print("Remaining missing values:")
display(df.isnull().sum())

print("\nRemaining duplicate records:")
print(df.duplicated().sum())

print("\nUpdated dataset shape:")
print(df.shape)

print("\nUpdated data types:")
display(df.dtypes)
```

Step 13: Statistical analysis of automobile specifications

```python
statistics_summary = df[numerical_columns].agg([
    "count",
    "mean",
    "median",
    "std",
    "min",
    "max"
]).T

statistics_summary = statistics_summary.round(2)

display(statistics_summary)
```

Step 14: Feature engineering – Power-to-weight ratio

```python
if all(
    col in df.columns
    for col in ["horsepower", "curb_weight"]
):

    df["power_to_weight"] = (
        df["horsepower"] / df["curb_weight"]
    )

    print("Power-to-weight ratio created.")

    display(
        df[
            [
                "horsepower",
                "curb_weight",
                "power_to_weight"
            ]
        ].head()
    )
```

Step 15: Feature engineering – Combined MPG

```python
if all(
    col in df.columns
    for col in ["city_mpg", "highway_mpg"]
):

    df["combined_mpg"] = (
        df["city_mpg"] + df["highway_mpg"]
    ) / 2

    print("Combined MPG created.")

    display(
        df[
            [
                "city_mpg",
                "highway_mpg",
                "combined_mpg"
            ]
        ].head()
    )
```

Step 16: Create additional automobile categories

```python
if "curb_weight" in df.columns:

    df["weight_category"] = pd.cut(
        df["curb_weight"],
        bins=[
            0,
            2000,
            3000,
            4000,
            np.inf
        ],
        labels=[
            "Light",
            "Medium",
            "Heavy",
            "Very Heavy"
        ]
    )

    display(
        df[
            ["curb_weight", "weight_category"]
        ].head()
    )
```

```python
if "engine_size" in df.columns:

    df["engine_category"] = pd.cut(
        df["engine_size"],
        bins=[
            0,
            100,
            200,
            300,
            np.inf
        ],
        labels=[
            "Small",
            "Medium",
            "Large",
            "Very Large"
        ]
    )

    display(
        df[
            ["engine_size", "engine_category"]
        ].head()
    )
```

Step 17: Analyze automobile manufacturers

```python
if "make" in df.columns:

    manufacturer_count = (
        df["make"]
        .value_counts()
        .reset_index()
    )

    manufacturer_count.columns = [
        "Manufacturer",
        "Vehicle_Count"
    ]

    display(manufacturer_count)
```

```python
if "make" in df.columns:

    plt.figure(figsize=(12, 6))

    sns.countplot(
        data=df,
        y="make",
        order=df["make"].value_counts().index
    )

    plt.title("Number of Automobile Records by Manufacturer")
    plt.xlabel("Number of Vehicles")
    plt.ylabel("Manufacturer")

    plt.show()
```

Step 18: Analyze fuel type distribution

```python
if "fuel_type" in df.columns:

    fuel_summary = (
        df["fuel_type"]
        .value_counts()
        .reset_index()
    )

    fuel_summary.columns = [
        "Fuel_Type",
        "Vehicle_Count"
    ]

    display(fuel_summary)

    plt.figure(figsize=(7, 5))

    sns.countplot(
        data=df,
        x="fuel_type"
    )

    plt.title("Automobile Distribution by Fuel Type")
    plt.xlabel("Fuel Type")
    plt.ylabel("Number of Vehicles")

    plt.show()
```

Step 19: Detect outliers using the IQR method

```python
outlier_columns = [
    "horsepower",
    "peak_rpm",
    "price",
    "engine_size",
    "curb_weight",
    "bore",
    "stroke"
]

outlier_columns = [
    col for col in outlier_columns
    if col in df.columns
]

outlier_results = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_results.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(2))
```

```python
for col in outlier_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[col])

    plt.title(f"Outlier Detection – {col}")
    plt.xlabel(col)

    plt.show()
```

Step 20: Correlation analysis

```python
correlation_columns = [
    "engine_size",
    "horsepower",
    "peak_rpm",
    "curb_weight",
    "city_mpg",
    "highway_mpg",
    "price",
    "power_to_weight",
    "combined_mpg"
]

correlation_columns = [
    col for col in correlation_columns
    if col in df.columns
]

correlation_matrix = df[
    correlation_columns
].corr()

display(correlation_matrix.round(2))

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Automobile Specifications")
plt.show()
```

```python
os.makedirs("week11_reports", exist_ok=True)

# Export cleaned automobile dataset
df.to_csv(
    "week11_reports/automobile_cleaned.csv",
    index=False
)

# Export statistical summary
statistics_summary.to_csv(
    "week11_reports/automobile_statistics_summary.csv"
)

# Export missing value summary
missing_summary.to_csv(
    "week11_reports/automobile_missing_value_summary.csv"
)

# Export outlier summary
outlier_summary.to_csv(
    "week11_reports/automobile_outlier_summary.csv",
    index=False
)

# Export correlation matrix
correlation_matrix.to_csv(
    "week11_reports/automobile_correlation_matrix.csv"
)

# Export manufacturer summary, if available
if "make" in df.columns:

    manufacturer_count.to_csv(
        "week11_reports/manufacturer_summary.csv",
        index=False
    )

# Export fuel summary, if available
if "fuel_type" in df.columns:

    fuel_summary.to_csv(
        "week11_reports/fuel_type_summary.csv",
        index=False
    )

print("All Week 11 reports exported successfully!")
```
