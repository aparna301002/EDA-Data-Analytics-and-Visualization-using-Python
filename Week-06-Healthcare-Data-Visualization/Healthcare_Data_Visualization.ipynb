# **Week 6: Healthcare Data Visualization and Exploratory Data Analysis using Python**

Step 1: Import required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully!")
```

Step 2: Load the cleaned healthcare dataset

```python
df = pd.read_csv("/content/drive/MyDrive/EDA-Skill/healthcare_preprocessed.csv")

df.head()
```

Step 3: Understand the dataset

```python
# Display first five records
print("First Five Records:")
display(df.head())

# Display last five records
print("\nLast Five Records:")
display(df.tail())

# Display dataset dimensions
print("\nDataset Shape:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display data types
print("\nData Types:")
print(df.dtypes)

# Display dataset information
print("\nDataset Information:")
df.info()
```

Step 4: Descriptive statistical analysis

```python
# Statistical summary of numerical columns

print("Numerical Statistical Summary:")
display(df.describe())

# Statistical summary of categorical columns

print("\nCategorical Statistical Summary:")
display(df.describe(include="object"))
```

Step 5: Univariate analysis

5.1 Age distribution

```python
plt.figure(figsize=(10, 5))

sns.histplot(df["age"], bins=20, kde=True)

plt.title("Distribution of Patient Age")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.show()
```

5.2 Gender distribution

5.4 Medical condition distribution

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="medical_condition",
    order=df["medical_condition"].value_counts().index
)

plt.title("Distribution of Medical Conditions")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)

plt.show()
```

5.5 Insurance provider distribution

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="insurance_provider",
    order=df["insurance_provider"].value_counts().index
)

plt.title("Distribution of Insurance Providers")
plt.xlabel("Insurance Provider")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)

plt.show()
```

Step 6: Bivariate analysis

6.1 Age vs Gender

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="gender",
    y="age"
)

plt.title("Age Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Age")

plt.show()
```

6.2 Medical condition vs Gender

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="medical_condition",
    hue="gender"
)

plt.title("Medical Conditions by Gender")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)

plt.legend(title="Gender")
plt.show()
```

6.3 Average billing by medical condition

6.4 Average billing by insurance provider

```python
insurance_billing = df.groupby(
    "insurance_provider"
)["billing_amount"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=insurance_billing.index,
    y=insurance_billing.values
)

plt.title("Average Billing Amount by Insurance Provider")
plt.xlabel("Insurance Provider")
plt.ylabel("Average Billing Amount")
plt.xticks(rotation=45)

plt.show()
```

6.5 Age vs Billing Amount

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="age",
    y="billing_amount",
    alpha=0.5
)

plt.title("Relationship Between Patient Age and Billing Amount")
plt.xlabel("Age")
plt.ylabel("Billing Amount")

plt.show()
```

Step 7: Multivariate analysis

```python
# Select numerical columns

numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation matrix

correlation_matrix = numeric_df.corr()

# Plot heatmap

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Healthcare Dataset")
plt.show()
```

7.2 Age group analysis

```python
# Create age groups

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 18, 35, 59, 120],
    labels=["0-18", "19-35", "36-59", "60+"]
)

# Count patients in each age group

age_group_counts = df["age_group"].value_counts().sort_index()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=age_group_counts.index,
    y=age_group_counts.values
)

plt.title("Patient Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")

plt.show()
```

7.3 Medical condition across age groups

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="age_group",
    hue="medical_condition"
)

plt.title("Medical Conditions Across Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")

plt.legend(title="Medical Condition", bbox_to_anchor=(1.05, 1))
plt.show()
```

Step 8: Admission date analysis

```python
# Convert admission date into datetime format

df["date_of_admission"] = pd.to_datetime(
    df["date_of_admission"],
    errors="coerce"
)

# Extract year and month

df["admission_year"] = df["date_of_admission"].dt.year
df["admission_month"] = df["date_of_admission"].dt.month

# Display admission date range

print("Earliest Admission Date:", df["date_of_admission"].min())
print("Latest Admission Date:", df["date_of_admission"].max())
```

8.1 Year-wise admission analysis

```python
yearly_admissions = df["admission_year"].value_counts().sort_index()

plt.figure(figsize=(10, 5))

sns.lineplot(
    x=yearly_admissions.index,
    y=yearly_admissions.values,
    marker="o"
)

plt.title("Year-wise Patient Admissions")
plt.xlabel("Admission Year")
plt.ylabel("Number of Admissions")

plt.xticks(yearly_admissions.index)
plt.show()
```

8.2 Month-wise admission analysis

```python
monthly_admissions = df["admission_month"].value_counts().sort_index()

month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

plt.figure(figsize=(10, 5))

sns.barplot(
    x=monthly_admissions.index,
    y=monthly_admissions.values
)

plt.title("Month-wise Patient Admissions")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")

plt.xticks(
    ticks=range(12),
    labels=month_names
)

plt.show()
```

Step 9: Outlier analysis

9.1 Billing amount boxplot

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["billing_amount"]
)

plt.title("Outlier Detection in Billing Amount")
plt.xlabel("Billing Amount")

plt.show()
```

9.2 Calculate outliers using IQR

```python
Q1 = df["billing_amount"].quantile(0.25)
Q3 = df["billing_amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["billing_amount"] < lower_bound) |
    (df["billing_amount"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Number of Potential Outliers:", len(outliers))
```

10.2 Billing amount by gender

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="gender",
    y="billing_amount"
)

plt.title("Billing Amount Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Billing Amount")

plt.show()
```

Step 11: Generate healthcare insights

```python
print("HEALTHCARE DATA ANALYSIS INSIGHTS")
print("----------------------------------")

# Total patient records
print("Total Patient Records:", len(df))

# Average age
print("Average Patient Age:", round(df["age"].mean(), 2))

# Most common gender
print("Most Common Gender:", df["gender"].mode()[0])

# Most common blood group
print("Most Common Blood Group:", df["blood_type"].mode()[0])

# Most common medical condition
print("Most Common Medical Condition:",
      df["medical_condition"].mode()[0])

# Most common insurance provider
print("Most Common Insurance Provider:",
      df["insurance_provider"].mode()[0])

# Average billing
print("Average Billing Amount:",
      round(df["billing_amount"].mean(), 2))

# Highest billing amount
print("Maximum Billing Amount:",
      df["billing_amount"].max())

# Lowest billing amount
print("Minimum Billing Amount:",
      df["billing_amount"].min())

# Year with maximum admissions
print("Year with Maximum Admissions:",
      df["admission_year"].value_counts().idxmax())
```

Step 12: Export analysis results

12.1 Export medical condition billing summary

```python
condition_summary = df.groupby(
    "medical_condition"
).agg(
    Total_Patients=("age", "count"),
    Average_Billing=("billing_amount", "mean"),
    Minimum_Billing=("billing_amount", "min"),
    Maximum_Billing=("billing_amount", "max")
).reset_index()

condition_summary.to_csv(
    "condition_billing_summary.csv",
    index=False
)

display(condition_summary)
```

12.3 Export insurance billing summary

```python
insurance_summary = df.groupby(
    "insurance_provider"
).agg(
    Total_Patients=("age", "count"),
    Average_Billing=("billing_amount", "mean"),
    Total_Billing=("billing_amount", "sum")
).reset_index()

insurance_summary.to_csv(
    "insurance_billing_summary.csv",
    index=False
)

display(insurance_summary)
```
